## Inspect data structure of EGMS data

```python
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path("/home/lukasscharf/projects/ADUCAT/data/EGMS/extracted")

csv_files = list(DATA_DIR.rglob("*.csv"))

print(f"Found {len(csv_files)} CSV files")
for f in csv_files:
    print(f)
```

```python
# get first csv file
csv_file = csv_files[0]

print(f"File: {csv_file.name}")
print(f"Path: {csv_file}")
print("=" * 80)

# Read only first 5 rows
df_sample = pd.read_csv(csv_file, nrows=5)

print(f"\nNumber of columns: {len(df_sample.columns)}")

print("\nColumn names:")
for i, col in enumerate(df_sample.columns):
    print(f"{i:2d}: {col}")

print("\nData types:")
print(df_sample.dtypes)

print("\nFirst 5 rows:")
display(df_sample)

print("\nDataFrame info:")
df_sample.info()
```
