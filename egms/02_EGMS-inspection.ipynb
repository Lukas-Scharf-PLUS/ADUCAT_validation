## Inspect data structure of EGMS data

```python
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path("/home/lukas/projects/ADUCAT/data/EGMS/extracted")

csv_files = list(DATA_DIR.rglob("*.csv"))

print(f"Found {len(csv_files)} CSV files")
for f in csv_files:
    print(f)
```

```python
# ============================================================
# Input settings
# ============================================================

ORBIT = "022"
PRODUCT_START = "2020"
PRODUCT_END = "2024"

# ============================================================
# Find input files
# ============================================================

PRODUCT_RANGE = f"{PRODUCT_START}_{PRODUCT_END}"

csv_files = sorted(
    f for f in DATA_DIR.glob("*.csv")
    if (
        f"_{ORBIT}_" in f.name
        and f"_{PRODUCT_RANGE}_" in f.name
    )
)

print(
    f"Found {len(csv_files)} CSV files "
    f"for orbit {ORBIT}, "
    f"product period {PRODUCT_RANGE}"
)

for f in csv_files:
    print(f"  {f.name}")
```

```python
# get first csv file
csv_file = csv_files[0]

print(f"File: {csv_file.name}")
print(f"Path: {csv_file}")
print("=" * 80)

# Read only first 5 rows
df_sample = pd.read_csv(csv_file, nrows=5)

print(f"\nNumber of columns: {len(df_sample.columns)}")

print("\nColumn names:")
for i, col in enumerate(df_sample.columns):
    print(f"{i:2d}: {col}")

print("\nData types:")
print(df_sample.dtypes)

print("\nFirst 5 rows:")
display(df_sample)

print("\nDataFrame info:")
df_sample.info()
```
